MACHINE LEARNING

Machine Learning es una rama de la Inteligencia Artificial que permite a las computadoras aprender a partir de datos y encontrar patrones para realizar predicciones o tomar decisiones sin ser programadas directamente para cada situación.

CLASIFICACIÓN

La clasificación es una técnica de Machine Learning que permite asignar datos a diferentes categorías o clases. Por ejemplo, clasificar un correo como spam o no spam. Puede ser clasificación binaria o multiclase.

DIAGNÓSTICO

El diagnóstico mediante Machine Learning consiste en analizar datos y patrones para identificar o predecir una determinada condición. Por ejemplo, utilizar síntomas y resultados médicos para clasificar si un paciente puede tener una enfermedad.

INTRODUCCIÓN A LOS DATOS

Los datos son la información utilizada para entrenar y evaluar un modelo de Machine Learning. Pueden contener características o variables de entrada y una etiqueta o resultado que se desea predecir. Los datos normalmente se dividen en datos de entrenamiento y datos de prueba.

CLASIFICADOR BINARIO

Un clasificador binario es un modelo que permite clasificar información en únicamente dos categorías. Por ejemplo: positivo o negativo, spam o no spam, enfermo o sano.

MATRIZ DE CONFUSIÓN

La matriz de confusión es una herramienta que permite evaluar un modelo de clasificación comparando las predicciones realizadas con los valores reales. Está formada por cuatro resultados: verdaderos positivos (TP), verdaderos negativos (TN), falsos positivos (FP) y falsos negativos (FN).

PRECISIÓN

La precisión indica qué tan correctas son las predicciones positivas realizadas por el modelo. Se calcula con la fórmula:

Precisión = TP / (TP + FP)

Una precisión alta significa que la mayoría de los casos clasificados como positivos realmente son positivos.

CURVA ROC

La curva ROC es una herramienta utilizada para evaluar un clasificador binario. Representa la relación entre la tasa de verdaderos positivos y la tasa de falsos positivos para diferentes valores de umbral. El área bajo la curva (AUC) permite medir el rendimiento del modelo; un valor cercano a 1 indica un buen clasificador y un valor cercano a 0.5 indica un rendimiento similar al azar.

CLASIFICACIÓN MULTICLASE

La clasificación multiclase se utiliza cuando un modelo debe elegir entre tres o más categorías. Por ejemplo, clasificar una imagen como perro, gato, caballo o pájaro. El modelo analiza las características de los datos y selecciona la clase que considera más adecuada.
