01 Entrenando Árboles: El proceso de entrenamiento construye el modelo de arriba hacia abajo mediante un enfoque codicioso (greedy). El algoritmo examina el conjunto de entrenamiento y busca iterativamente la característica y el umbral de corte específicos que mejor separen los datos en subconjuntos más definidos. Este proceso se repite de forma recursiva en cada nuevo nodo hasta que se alcanza una regla de detención predefinida.

02 Predicciones y Probas (Probabilidades): Para clasificar una nueva observación, sus datos ingresan por el nodo raíz y descienden respondiendo a las condiciones lógicas de cada nodo interno hasta aterrizar en un nodo hoja final. La predicción resultante es la clase dominante entre las muestras de entrenamiento que residen en esa hoja. La probabilidad de pertenencia a una clase se extrae calculando simplemente la proporción de muestras de esa clase específica dentro del nodo hoja alcanzado.

03 CART e Impureza: CART (Classification and Regression Trees) es el algoritmo estándar empleado por bibliotecas como Scikit-Learn para crear estos modelos, caracterizado por generar siempre árboles binarios (divisiones de verdadero o falso). Para decidir cuál es la mejor división en cada paso, CART evalúa funciones de costo basadas en la "impureza", como la Impureza de Gini o la Entropía. El objetivo matemático en cada corte es minimizar esta impureza, garantizando que los nodos resultantes contengan clases lo más puras u homogéneas posible.

04 Regularización: Por naturaleza, un árbol de decisión sin restricciones crecerá hasta memorizar cada anomalía y ruido del conjunto de entrenamiento, un problema crítico conocido como sobreajuste (overfitting). La regularización mitiga esto aplicando restricciones estructurales durante el entrenamiento. Esto se logra ajustando hiperparámetros de límite, como definir una profundidad máxima para el árbol, exigir un número mínimo de muestras para permitir que un nodo se divida, o requerir un volumen mínimo de datos para conformar una hoja final.

05 Árboles de Regresión: Esta es la adaptación del algoritmo para predecir variables continuas (números reales) en lugar de categorías. La mecánica de división es idéntica, pero el criterio de evaluación cambia: en lugar de buscar la pureza de clases mediante Gini o entropía, el algoritmo de regresión busca cortes que minimicen el Error Cuadrático Medio (MSE) o la varianza. El valor final predicho por el modelo será el promedio numérico de todas las muestras de entrenamiento que cayeron en ese nodo hoja.

06 Limitaciones: La principal vulnerabilidad de los árboles de decisión es su inestabilidad geométrica; variaciones mínimas en los datos de entrenamiento o la eliminación de unas pocas muestras pueden resultar en una arquitectura de árbol completamente distinta. Adicionalmente, construyen límites de decisión estrictamente ortogonales (escalonados), lo que los hace ineficientes frente a conjuntos de datos rotados o patrones diagonales. Por estas debilidades, en entornos de producción los árboles rara vez se utilizan solos, funcionando más a menudo como los bloques base para métodos de ensamble avanzados como Random Forests o Gradient Boosting.
